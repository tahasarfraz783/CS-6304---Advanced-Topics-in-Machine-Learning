Known training features only. Pooled within-class diagonal covariance uses maximum-likelihood divisor N and adds 1e-6.

```python
def fit_mahalanobis(f,y):
    f=np.asarray(f,dtype=np.float64);y=np.asarray(y)
    if set(np.unique(y))!=set(range(10)):raise ValueError('All known classes are required.')
    means=np.stack([f[y==c].mean(0) for c in range(10)])
    variance=((f-means[y])**2).mean(0)+1e-6
    return means,variance

def mahalanobis(f,means,variance):
    f=np.asarray(f,dtype=np.float64)
    # Process one class at a time instead of materializing N x 10 x 512.
    return np.stack([np.sum((f-mu)**2/variance,axis=1) for mu in means],axis=1).min(1)
```
