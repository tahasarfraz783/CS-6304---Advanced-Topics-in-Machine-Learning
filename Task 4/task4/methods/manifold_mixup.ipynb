Mix after layer2; shuffled same-class pairs are excluded, following the paper. One Beta(2,2) draw per batch.

```python
def mix_hidden(h,y):
    permutation=torch.randperm(len(y),device=y.device)
    valid=y!=y[permutation]
    lam=float(np.random.beta(2.,2.))
    return lam*h[valid]+(1.-lam)*h[permutation][valid],int(valid.sum().item())
```
