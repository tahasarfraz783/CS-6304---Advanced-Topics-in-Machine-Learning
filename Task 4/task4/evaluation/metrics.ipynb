AUROC uses unknown=1. FPR is unknowns accepted at a threshold calibrated to 95% known validation acceptance.

```python
from sklearn.metrics import roc_auc_score,roc_curve

def osr_metrics(known,near,far,tau):
    result={'threshold':float(tau),'known_test_acceptance':float(np.mean(known<=tau))}
    for name,u in [('near',near),('far',far),('all',np.concatenate([near,far]))]:
        result['auroc_'+name]=float(roc_auc_score(np.r_[np.zeros(len(known)),np.ones(len(u))],np.r_[known,u]))
        result['rejection_'+name]=float(np.mean(u>tau))
        result['fpr95val_'+name]=float(np.mean(u<=tau))
    return result
```
